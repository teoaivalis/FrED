Check Artists

```python
# Group by artist and count the number of images
artist_counts = df.groupby('artist').size().reset_index(name='painting_count')
# Sort them by the number of paintings
artist_counts = artist_counts.sort_values(by='painting_count', ascending=False)
```

```python
print(f"Total unique artists in training set: {len(artist_counts)}")
```

```python
print("Paintings per Artist:")
print(artist_counts.to_string(index=False))
```

Focus on Artbench 2 (Post Impressionism & Ukiyo-e)

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

train_df = df[df['split'] == 'train'].copy()

styles_of_interest = ['post_impressionism', 'ukiyo_e']

print(f"Total training samples: {len(train_df)}")
for style in styles_of_interest:
    count = len(train_df[train_df['label'] == style])
    print(f"Samples in {style}: {count}")
```

```python
for style in styles_of_interest:
    style_data = train_df[train_df['label'] == style]
    artist_counts = style_data['artist'].value_counts()
    
    print(f"\n=== Statistics for {style.upper()} ===")
    print(f"Total Unique Artists: {len(artist_counts)}")
    print(f"Top 5 Artists and their training counts:")
    print(artist_counts)
```
