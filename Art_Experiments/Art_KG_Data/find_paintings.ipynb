```python
import pandas as pd

# 1. Load your initial dataframe (ArtBench)
# df = pd.read_csv('ArtBench-10.csv') # Uncomment and load your file
# Assuming 'df' is already loaded as per previous steps

# 2. Load the metadata CSV (the one with 'tags', 'genre', etc.)
df_csv = pd.read_csv('wikiart_art_pieces.csv')

print("Searching for matches and combining info...\n")
matches_found = 0

# Iterate through the original dataframe 'df'
for index, row in df.iterrows():
    link = row['url']
    
    # Search for this link in the 'img' column of the metadata CSV
    # We use .str.contains() because the CSV links have '!Large.jpg' appended
    match = df_csv[df_csv['img'].str.contains(link, regex=False)]
    
    if not match.empty:
        matches_found += 1
        
        # Get the first matching row from the CSV
        csv_row = match.iloc[0]
        
        print(f"--- Record {matches_found} ---")
        
        # --- Info from the CSV File ---
        print("From CSV File:")
        print(f"  Artist:   {csv_row['artist']}")
        print(f"  Style:    {csv_row['style']}")
        print(f"  Genre:    {csv_row['genre']}")
        print(f"  Movement: {csv_row['movement']}")
        print(f"  Tags:     {csv_row['tags']}")
        
        # --- Info from the DataFrame (ArtBench) ---
        print("From DataFrame:")
        print(f"  Name:     {row['name']}")
        print(f"  Artist:   {row['artist']}")
        print(f"  URL:      {row['url']}")
        
        print("-" * 30)
        
        if matches_found >= 5:
            break

if matches_found == 0:
    print("No matches found.")
```

```python
import pandas as pd
import json

# 1. Load the Dataframes
# Replace with your actual file paths
artbench_path = '/home/teoaivalis/.cache/kagglehub/datasets/alexanderliao/artbench10/versions/2/ArtBench-10.csv'
metadata_path = 'wikiart_art_pieces.csv'

df_artbench = pd.read_csv(artbench_path)
df_metadata = pd.read_csv(metadata_path)

print("Files loaded. Processing...")

# 2. Prepare the Metadata for Merging
# The metadata 'img' column has '!Large.jpg' at the end.
# We create a temporary column 'clean_url' to match the ArtBench 'url'.
df_metadata['clean_url'] = df_metadata['img'].str.replace('!Large.jpg', '', regex=False)

# 3. Merge the Dataframes
# This combines rows where ArtBench['url'] matches Metadata['clean_url']
# how='inner' keeps only rows that exist in BOTH files.
merged_df = pd.merge(
    df_artbench, 
    df_metadata, 
    left_on='url', 
    right_on='clean_url', 
    how='inner'
)

# 4. Select and Rename the specific columns you want
# Note: When columns have the same name (like 'artist'), pandas adds _x (left/ArtBench) and _y (right/Metadata)
final_df = merged_df[[
    # From ArtBench (df)
    'name', 
    'artist_x',    # This is the artist from ArtBench
    'url_x',       # This is the url from ArtBench
    
    # From Metadata (csv)
    'artist_y',    # This is the artist from Metadata
    'style', 
    'genre', 
    'movement', 
    'tags'
]]

# Rename them for the final JSON to be clean
final_df.columns = [
    'name', 
    'artbench_artist', 
    'url', 
    'metadata_artist', 
    'style', 
    'genre', 
    'movement', 
    'tags'
]

# 5. Save to JSON
output_filename = 'artbench_full_metadata.json'
print(f"Merged {len(final_df)} records. Saving to {output_filename}...")

# orient='records' creates a list of objects: [{...}, {...}]
json_data = final_df.to_dict(orient='records')

with open(output_filename, 'w') as f:
    json.dump(json_data, f, indent=4)

print("Done! File saved successfully.")
```
